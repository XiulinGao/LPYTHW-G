```python
## list: indexable and mutable 

fruit = [
    ['Apples', 12, 'AAA'], ['Oranges', 1, 'B'],
    ['Pears', 2, 'A'], ['Grapes', 14, 'UR']]

cars = [
    ['Cadillac', ['Black', 'big', 34500]],
    ['Corvette', ['Red', 'Little', 1000000]],
    ['Ford', ['Blue', 'Medium', 1234]],
    ['BMW', ['White', 'Baby', 7890]]
]

lang = [
    ['Python', ['Slow', ['Terrible', 'Mush']]],
    ['JavaScript', ['Moderate', ['Alright', 'Bizarre']]],
    ['Perl6', ['Moderate', ['Fun', 'Weird']]],
    ['C', ['Fast', ['Annoying', 'Dangerous']]],
    ['Forth', ['Fast', ['Fun', 'Difficult']]]
]
```

```python
## list indexing
print(fruit[0][1]) # 12
print(fruit[0][2]) # AAA
print(fruit[2][1]) # 2
print(fruit[1][0]) # oranges
print(fruit[3][0]) # grapes
print(fruit[3][1]) # 14
print(fruit[0][0]) # apples
```

```python
print(f"*{cars[0][1][1]}\n*{cars[1][1][0]}\n*{cars[2][1][2]}\n*{cars[3][1][0]}\n*{cars[3][1][2]}\n*{cars[0][1][0]}\n*{cars[0][1][2]}\n*{cars[2][1][0]}")
```

```python
print(f"*{lang[0][1][0]}\n*{lang[1][1][1][0]}\n*{lang[3][1][1][1]}\n*{lang[3][1][0]}\n*{lang[4][1][1][1]}\n*{lang[2][1][1][0]}\n*{lang[3][1][1][0]}\n*{lang[2][1][1][1]}\n*{lang[1][1][0]}")
```

```python
print(f"*{cars[1][1][1]}\n*{cars[1][1][0]}\n*{cars[1][0]}\n*{cars[3][1][1]}\n*{fruit[3][2]}\n*{lang[0][1][1][1]}\n*{fruit[2][1]}\n*{lang[3][1][0]}") #little red Corvette Baby UR Mush 2 Fast
```

```python
## Dictionaries: mutable but not indexable; as values are positioned by key 

## pure dict
email = {
    "From": "j.smith@example.com",
    "TO": "well.done@example.com",
    "Subject": "I have an amazing investment for you!"
}

print(email["TO"])
print(email["Subject"])

## combination of list and dict
messages = [
    {"to": 'sun', "from": 'moon', "message": 'Hi!'},
    {"to": 'moon', "from": 'sun', "message": 'What do you want sun?'},
    {"to": 'sun', "from": 'moon', "message": "I'm awake!"},
    {"to": 'moon', "from": 'sun', "message": 'I can see the sun.'}
]
print(messages[0]['message'])
print(messages[1]['message'])
print(messages[2]['from'])
```

```python
fruit = [
    {'kind': 'Apples',  'count': 12, 'rating': 'AAA'},
    {'kind': 'Oranges', 'count': 1,  'rating': 'B'},
    {'kind': 'Pears',   'count': 2,  'rating': 'A'},
    {'kind': 'Grapes',  'count': 14, 'rating': 'UR'}
];

cars = [
    {'type': 'Cadillac', 'color': 'Black',
     'size': 'Big', 'miles': 34500},
    {'type': 'Corvette', 'color': 'Red',
     'size': 'Little', 'miles': 1000000},
    {'type': 'Ford', 'color': 'Blue',
     'size': 'Medium', 'miles': 1234},
    {'type': 'BMW', 'color': 'White',
     'size': 'Baby', 'miles': 7890}
];

lang = [
    {'name': 'Python', 'speed': 'Slow',
     'opinion': ['Terrible', 'Mush']},
    {'name': 'JavaScript', 'speed': 'Moderate',
     'opinion': ['Alright', 'Bizarre']},
    {'name': 'Perl6', 'speed': 'Moderate',
     'opinion': ['Fun', 'Weird']},
    {'name': 'C', 'speed': 'Fast',
     'opinion': ['Annoying', 'Dangerous']},
    {'name': 'Forth', 'speed': 'Fast',
     'opinion': ['Fun', 'Difficult']},
];
```

```python
print(f"*{fruit[0]['count']}\n*{fruit[0]['rating']}\n*{fruit[2]['count']}\n*{fruit[1]['kind']}\n*{fruit[3]['kind']}\n*{fruit[3]['count']}\n*{fruit[0]['kind']}")
print("==================")
print(f"*{cars[0]['size']}\n*{cars[1]['color']}\n*{cars[2]['miles']}\n*{cars[3]['color']}\n*{cars[3]['miles']}\n*{cars[0]['color']}\n*{cars[0]['miles']}\n*{cars[2]['color']}")
print("==================")
print(f"*{lang[0]['speed']}\n*{lang[1]['opinion'][0]}\n*{lang[3]['opinion'][1]}\n*{lang[3]['speed']}\n{lang[4]['opinion'][1]}\n*{lang[2]['opinion'][0]}\n*{lang[3]['opinion'][0]}\n*{lang[2]['opinion'][1]}\n*{lang[1]['speed']}")
print(f"{cars[1]['size']} {cars[1]['color']} {cars[1]['type']} {cars[3]['size']} {fruit[3]['rating']} {lang[0]['opinion'][1]} {fruit[2]['count']} {lang[3]['speed']}")
```
