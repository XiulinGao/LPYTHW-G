```python
## let's create a fighting club game!
import random as rd
from sys import exit


def smash_bros(name, age, eyes, points, role):
  
    point_system = {
        "knight": {"hp": 5000, 
                   "damage": 50,
                  "dialog": "Zelda, I'm coming for you!"},
        "magician": {"hp": 1000,
                     "damage": 100,
                    "dialog": "Link, wake up..."},
        "demon": {"hp": 10000,
                  "damage": 100,
                  "dialog": "I AM the king of Hyrule!"},
        "armed_knight": {"hp": 5000,
                         "damage": 150,
                         "dialog": "Ganondorf is a coward."}
    }

    role_info = point_system[role]

    person = {
        "name": name,
        "age": age,
        "eyes": eyes,
        "points": points,
        "role": role,
        "hp": role_info['hp'],
        "damage": role_info['damage'],
        "dialog": role_info['dialog']
    }

    

    def talk():
        print(f"{person['name']}: {person['dialog']}")
    
    person['talk'] = talk
    
    def hit(p1, p2):
        if p1['hp'] <= 0 or p2['hp'] <= 0:
             exit("💀 One of the players is already dead. Exiting game.") 
        print(f"\n⚔️  {p1['name']} is fighting with {p2['name']}")
        p1['talk']()
        p2['talk']()
        damage = p1['damage']
        p1['points'] += damage
        p2['hp'] -= damage
        print(f"{p1['name']} earned {damage} points (Total: {p1['points']}")
        print(f"{p2['name']} lost {damage} hp (Remaining: {p2['hp']}") 
        if p2['hp'] <= 0:
            print(f"💀 {p2['name']} has died!")
            exit("Game over!")
    
    person['hit'] = hit   
    return person
               
   
    
link = smash_bros("Link", 100, "blue", 0, "knight")
zelda = smash_bros("Zelda", 100, "blue", 0, "magician")
ganon = smash_bros("Ganondorf", 500, "red", 0, "demon")
master_link = smash_bros("Master_link", 100, "blue", 0, "armed_knight")

players = {
    "Link": link,
    "Zelda": zelda,
    "Ganondorf": ganon,
    "Master_link": master_link
}

player_list = list(players.keys())

print("🕹️ Welcome to Smash Bros Arena!")
print("Available players:", player_list)

try:
    num = int(input("🎮 How many rounds do you want to play? "))
except ValueError:
    exit("❌ Please enter a valid number.")

p1 = input("👤 Choose your main player (case-sensitive): ")
if p1 not in players:
    exit(f"❌ Invalid player name: {p1}. Exiting.")
    
for rep in range(num):
     remains = [name for name in player_list if name != p1]
     p2 = rd.choice(remains)
    
     player1 = players[p1]
     player2 = players[p2]
     hit_now = player1['hit']
     hit_now(player1, player2)
```
