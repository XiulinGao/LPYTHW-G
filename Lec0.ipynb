```python
cars = 100
space_in_car = 4.0
drivers = 30
passengers = 90
cars_not_driven = cars - drivers
cars_driven = drivers
carpool_capacity = cars_driven*space_in_car
average_passengers_per_car = passengers/cars_driven

print("There are ", cars, "cars available.")
print("There are only ", drivers, "drivers available.")
print("There will be ", cars_not_driven, "empty cars.")
print("We can transport ", carpool_capacity, "passengers today.")
print("We have ", passengers, "people need ride today.")
print("We need to put about ", average_passengers_per_car, "people in each car.")
```

```python
## use \ as escape sequence
print("I'm 5'6\" tall") #double quote
print('I\'m 5\'6" tall') #single quote

tabby_cat = "\tI'm tabbed in." # \t indicates a tab 
persian_cat = "I'm split \non a line." # \n start a new line after split
backslash_cat = "I'm \\a\\ cat." # tell python that \a\ are needed to print out. Otherwise it will output I'm \ cat as a is escaped 

fat_cat = """
I'll do a list.
\t* Cat food
\t* Fishies
\t* Catnip\n\t* Grass
"""

fat_cat2 = """
I'll do a list.
\t* Cat food\n\t* Fishes\n\t* Catnip\n\t* Grass
"""

print(tabby_cat)
print(persian_cat)
print(backslash_cat)
print(fat_cat)
print(fat_cat2)
```
