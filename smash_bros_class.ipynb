```python
## let's create a fighting club game!
import random as rd
from sys import exit


class SmashBros(object):
    role_info = {
        "knight": {"hp": 5000, 
                   "damage": 50,
                  "dialog": "Zelda, I'm coming for you!"},
        "magician": {"hp": 1000,
                     "damage": 100,
                    "dialog": "Link, wake up..."},
        "demon": {"hp": 10000,
                  "damage": 100,
                  "dialog": "I AM the king of Hyrule!"},
        "armed_knight": {"hp": 5000,
                         "damage": 150,
                         "dialog": "Ganondorf is a coward."}
     }
    
    def __init__(self, name, age, eyes, points, role):
        self.name = name
        self.age  = age
        self.eyes = eyes
        self.points = points
        self.role = role
        self.hp = smash_bros.role_info[role]['hp']
        self.damage = smash_bros.role_info[role]['damage']
        self.dialog = smash_bros.role_info[role]['dialog']
          
    def talk(self):
        print(f"{self.name}: {self.dialog}")

    
    def hit(self, other):
        if self.hp <= 0 or other.hp <= 0:
             exit("💀 One of the players is already dead. Exiting game.") 
        print(f"\n⚔️  {self.name} is fighting with {other.name}")
        self.talk()
        other.talk()
        self.points += self.damage
        other.hp -= self.damage
        print(f"{self.name} earned {self.damage} points (Total: {self.points})")
        print(f"{other.name} lost {self.damage} hp (Remaining: {other.hp})") 
        if other.hp <= 0:
            print(f"💀 {other.name} has died!")
            exit("Game over!")
               
      
link = SmashBros("Link", 100, "blue", 0, "knight")
zelda = SmashBros("Zelda", 100, "blue", 0, "magician")
ganon = SmashBros("Ganondorf", 500, "red", 0, "demon")
master_link = SmashBros("Master_link", 100, "blue", 0, "armed_knight")

players = {
    "Link": link,
    "Zelda": zelda,
    "Ganondorf": ganon,
    "Master_link": master_link
}

player_list = list(players.keys())

print("🕹️ Welcome to Smash Bros Arena!")
print("Available players:", player_list)

try:
    num = int(input("🎮 How many rounds do you want to play? "))
except ValueError:
    exit("❌ Please enter a valid number.")

p1 = input("👤 Choose your main player (case-sensitive): ")
if p1 not in players:
    exit(f"❌ Invalid player name: {p1}. Exiting.")
    
for rep in range(num):
     remains = [name for name in player_list if name != p1]
     p2 = rd.choice(remains)
    
     player1 = players[p1]
     player2 = players[p2]
     player1.hit(player2)
```
