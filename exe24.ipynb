```python
## dicts with functions

def run():
    print("VROOM")

corvette = {
    "color": "Red",
    "run": run
}

print("My", corvette["color"], "can go")
corvette["run"]()
```

```python
## dicts and modules
## import content in exe26 as a module

import exe26
print("name", exe26.name)
print("height", exe26.height)
```

```python
from pprint import pprint  #pretty printer 
pprint(exe26.__dict__)
```

```python
print("height is: ", exe26.height)
print("height is: ", exe26.__dict__['height'])
```

```python
## if a module is really a dict inside, then you can change the contents of __dict__ in the module

print(f"I am currently {exe26.height} inches tall.")
exe26.__dict__['height'] = 1000
print(f"I am currently {exe26.height} inches tall.")
exe26.height = 12
print(f"I am currently {exe26.height} inches tall.")
```

```python
## if you change exe26.py and need to reload running import again won't work.
## as Python "caches" modules to make them faster. to reload you 
## need to use the import.reload function
## one caveat is that Python will keep variables in exe26.py even if you reload. 
## so if you delete a variable the look at the __dict__ you will still see it. 

import exe26
print(exe26.height) # not showing the new value after changing height

from importlib import reload
reload(exe26)
print(exe26.height)
```
