```python
## let's create a fighting club game!
import random as rd
from sys import exit


def smash_bros(name, age, eyes, points, role):
  
    point_system = {
        "knight": {"hp": 5000, 
                   "damage": 50,
                  "dialog": "Zelda, I'm coming for you!"},
        "magician": {"hp": 1000,
                     "damage": 100,
                    "dialog": "Link, wake up..."},
        "demon": {"hp": 10000,
                  "damage": 100,
                  "dialog": "I AM the king of Hyrule!"},
        "armed_knight": {"hp": 5000,
                         "damage": 150,
                         "dialog": "Ganondorf is a coward."}
    }

    role_info = point_system[role]

    person = {
        "name": name,
        "age": age,
        "eyes": eyes,
        "points": points,
        "role": role,
        "hp": role_info['hp'],
        "damage": role_info['damage'],
        "dialog": role_info['dialog']
    }

    

    def talk():
        print(f"{person['name']}: {point_system[person['role']]['dialog']}")
    
    person['talk'] = talk
    
    def hit(p1, p2):
        if p1['hp'] <= 0 or p2['hp'] <= 0:
            exit("At least of the players is dead, exit game!") 
        print(f"{p1['name']} is fighting with {p2['name']}")
        p1['talk']()
        p2['talk']()
        damage = p1['damage']
        p1['points'] += damage
        p2['hp'] -= damage
        print(f"{p1['name']} earned {damage} points")
        print(f"{p2['name']} lost {damage} hp and has {p2['hp']} left.")  
    
    person['hit'] = hit   
    return person
               
   
    
link = smash_bros("Link", 100, "blue", 0, "knight")
zelda = smash_bros("Zelda", 100, "blue", 0, "magician")
ganon = smash_bros("Ganondorf", 500, "red", 0, "demon")
master_link = smash_bros("Master_link", 100, "blue", 0, "armed_knight")

players = {
    "Link": link,
    "Zelda": zelda,
    "Ganondorf": ganon,
    "Master_link": master_link
}

player_list = list(players.keys())

print("please input the number of fights: ")
num = int(input( ))
print("please input the main player name (choose from):", player_list)
p1 = input()

if p1 not in players:
    exit("Invalid player name!")
    
for rep in range(num):
     remains = player_list.copy()
     remains.remove(p1)
     p2 = rd.choice(remains)
    
     player1 = players[p1]
     player2 = players[p2]
     hit_now = player1['hit']
     hit_now(player1, player2)
```
